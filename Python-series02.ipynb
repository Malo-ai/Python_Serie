```python
# Import needed libraries
import numpy as np
import matplotlib.pyplot as plt

# Code cells
t = np.linspace(0, 1, 1000)  # time vector from 0 to 1 second with 1000 samples
s = np.sin(2 * np.pi * 1 * t)  # sinusoidal function
c = np.cos(2 * np.pi * 1 * t)  # cosine function
plt.plot(t, s)          # plot the function
plt.plot(t, c)          # plot the cosine function
plt.title('Sinusoidal Wave')  # title
plt.legend(['sinus)', 'cosinus)'])  # legend
plt.xlabel('Time (s)')        # x-axis label
plt.ylabel('Amplitude')       # y-axis label
plt.grid(True)               # add grid

# Save the plot as a PDF file
plt.savefig('Sinusoidal_Wave.pdf', bbox_inches='tight')  # Save the plot as a PDF f

plt.show()                  # display the plot
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(amplitude, frequency, phase, duration, sampling_rate):
    """
    Generate a sinusoidal signal.

    Parameters
    ----------
    amplitude : float
        Peak value of the sinusoid
    frequency : float
        Frequency in Hz
    phase : float
        Phase shift in radians
    duration : float
        Duration of the signal in seconds
    sampling_rate : float
        Number of samples per second

    Returns
    -------
    numpy.ndarray
        The generated sinusoidal signal
    """
    t = np.linspace(0, duration, int(sampling_rate*duration), endpoint=False)
    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x, t

# Parameters
amp = 1.0
freq1 = 1  # Hz
freq2 = 2 # Hz
phase = 0
duration = 1.0 # seconds
fs = 1000      # sampling rate in Hz

# Generate signals
s1, t = sinusoid(amp, freq1, phase, duration, fs)
s2, _ = sinusoid(amp, freq2, phase, duration, fs)

# Plot signals
plt.figure(figsize=(10,5))
plt.plot(t, s1, label=f'{freq1} Hz')
plt.plot(t, s2, label=f'{freq2} Hz')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.title('Sinusoidal Signals')
plt.legend()
plt.grid(True)
plt.show()

# References:
# - NumPy documentation: https://numpy.org/doc/stable/reference/generated/numpy.sin.html
# - Matplotlib documentation: https://matplotlib.org/stable/tutorials/introductory/pyplot.html
# - Signal processing principles: Oppenheim, A.V. & Schafer, R.W., Discrete-Time Signal Processing, 3rd Edition
```
