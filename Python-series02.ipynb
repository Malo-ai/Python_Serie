```python
# Import libraries (once)
import numpy as np
import matplotlib.pyplot as plt

# Part 1: Simple sine and cosine
t = np.linspace(0, 1, 1000)        # time vector
s = np.sin(2 * np.pi * t)          # sine
c = np.cos(2 * np.pi * t)          # cosine

plt.figure(figsize=(10,5))
plt.plot(t, s, label='sine', linewidth=2)   # thicker line
plt.plot(t, c, label='cosine', linewidth=2)
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # baseline at y=0
plt.title('Sinusoidal Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Save plot
plt.savefig('Sinusoidal_Wave.pdf', bbox_inches='tight')  # Save the plot as a PDF file
plt.savefig('C:\\Users\\malob\\Desktop\\IEAP\\Python_Serie\\Python_Serie_02\\Export\\Sinusoidal_Wave.pdf', bbox_inches='tight')
plt.show()
```

# Part 2: Multiple sinusoids (general)

```python
# Inline sinusoid function
def sinusoid(amplitude, frequency, phase, duration, sampling_rate):
    t = np.linspace(0, duration, int(sampling_rate*duration), endpoint=False)
    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x, t

# Parameters
amp = 1.0
freqs = [1, 2]       # Hz
phase = 0            # radians  
duration = 1.0       # seconds
fs = 1000            # sampling rate

# Generate signals
signals = [sinusoid(amp, f, phase, duration, fs)[0] for f in freqs]
t = sinusoid(amp, freqs[0], phase, duration, fs)[1]

# Plot
plt.figure(figsize=(10,5))
for sig, f in zip(signals, freqs):
    plt.plot(t, sig, label=f'{f} Hz', linewidth=2)
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # baseline
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.title('Sinusoidal Signals')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# References:
# - NumPy documentation: https://numpy.org/doc/stable/reference/generated/numpy.sin.html
# - Matplotlib documentation: https://matplotlib.org/stable/tutorials/introductory/pyplot.html
# - Signal processing principles: Oppenheim, A.V. & Schafer, R.W., Discrete-Time Signal Processing, 3rd Edition
```
